**46) Create a variable `fav_number` equal to your favorite integer. Write a loop to determine if `fav_number` is prime.**

````python
fav_number = 17
is_prime = True
if fav_number < 2:
    is_prime = False
else:
    for i in range(2, fav_number):
        if fav_number % i == 0:
            is_prime = False
            break
print(is_prime)
````

**47) A palindromic number reads the same both ways. Write an `if` statement to test whether or not a number is palindromic.**

````python
number = 1234321
if str(number) == str(number)[::-1]:
    print(str(number) + " is a palindrome")
else:
    print(str(number) + " is not a palindrome")
````

**48) Find the largest palindrome made from the product of two two-digit numbers.**

````python
largest = 0
for i in range(10, 100):
    for j in range(10, 100):
        product = i * j
        if str(product) == str(product)[::-1] and product > largest:
            largest = product
largest
````
