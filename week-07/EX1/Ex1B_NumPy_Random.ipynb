## Generating Random Numbers

```python
# r_1: random integers between 1 and 20, size = 2 elements (1D)
r_1 = np.random.randint(1, 20, size=2)
print(r_1)
```

```python
# r_2: random integers between 1 and 20, size = 1 row, 2 columns (2D)
r_2 = np.random.randint(1, 20, size=(1, 2))
print(r_2)
```

```python
# r_3: random floats between 0 and 1, size = 2 rows, 1 column
r_3 = np.random.random(size=(2, 1))
print(r_3)
```

```python
# r_4: random integers between 20 and 100, size = 4 rows, 5 columns
r_4 = np.random.randint(20, 100, size=(4, 5))
print(r_4)
```

```python
# r_5: random floats between 0 and 1, size = 4 rows, 5 columns
r_5 = np.random.random(size=(4, 5))
print(r_5)
```

## Generating Random Choices from an Array

```python
# Create a 1D array of 12 classmate names
students = np.array([
    "Kidus", "Aisha", "Marcus", "Fatima",
    "Jordan", "Leila", "Darius", "Amara",
    "Tyler", "Zara", "Elias", "Nadia"
])
```

```python
# call_on: choose ONE random student
call_on = np.random.choice(students)
print(call_on)
```

```python
# breakout_groups: random sample from students, 3 rows x 4 columns
breakout_groups = np.random.choice(students, size=(3, 4))
print(breakout_groups)

# Would this be an effective way to assign breakout groups?
# NOT really — np.random.choice() samples WITH replacement by default,
# meaning the same student could appear more than once in the grid.
# A real breakout group assignment needs each student to appear exactly once.
# A better approach would be np.random.shuffle() or np.random.choice(students, replace=False)
```

```python
# call_on_weighted: 5 random selections with custom probability weights
# Probability assignments (must add up to 1.0):
#   Kidus   = 0.30  (30% — most likely to be called on, the star student!)
#   Aisha   = 0.15
#   Marcus  = 0.10
#   Fatima  = 0.10
#   Jordan  = 0.10
#   Leila   = 0.08
#   Darius  = 0.07
#   Amara   = 0.05
#   Tyler   = 0.03
#   Zara    = 0.02
#   Elias   = 0.00  (0% — on vacation, never gets called on)
#   Nadia   = 0.00  (0% — also never called on)

call_on_weighted = np.random.choice(
    students,
    size=5,
    p=[0.30, 0.15, 0.10, 0.10, 0.10, 0.08, 0.07, 0.05, 0.03, 0.02, 0.00, 0.00]
)
print(call_on_weighted)
```

## Generating a Normal Distribution

```python
# bell_curve: normal distribution
# mean (loc) = 10  → peak of the bell curve is at 10
# std  (scale) = 5 → spread of the curve, 68% of values fall within 5 of the mean
# size = 4 rows, 5 columns
bell_curve = np.random.normal(loc=10, scale=5, size=(4, 5))
print(bell_curve)
```
