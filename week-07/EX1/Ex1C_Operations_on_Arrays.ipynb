## Sample Arrays

```python
# Copying sample arrays from Exercise 1.B
r_1 = np.random.randint(1, 20, size=2)
r_2 = np.random.randint(1, 20, size=(1, 2))
r_3 = np.random.random(size=(2, 1))
r_4 = np.random.randint(20, 100, size=(4, 5))
r_5 = np.random.random(size=(4, 5))

print("r_1:", r_1)
print("r_2:", r_2)
print("r_3:", r_3)
print("r_4:", r_4)
print("r_5:", r_5)
```

## Permutations

```python
# Print r_4 before shuffling
print("r_4 before shuffle:")
print(r_4)
```

```python
# Shuffle r_4 in place
np.random.shuffle(r_4)
print("r_4 after shuffle:")
print(r_4)
```

### What happened when we shuffled?

`np.random.shuffle()` shuffled the **rows** of `r_4`, not individual elements.
The values within each row stayed together — only the order of the rows changed.
This is because shuffle operates along the **first axis (axis=0)** only.

```python
# Create new_shuffle using permutation (does NOT modify r_4)
new_shuffle = np.random.permutation(r_4)
print("new_shuffle:")
print(new_shuffle)
print("\nr_4 (unchanged):")
print(r_4)
```

### shuffle vs permutation — when to use each?

Use **`shuffle`** when you want to rearrange the original array in place and don't need
to keep the original order. It modifies the array directly — no new variable needed.

Use **`permutation`** when you want a shuffled copy but need to preserve the original array.
It returns a new array and leaves the original untouched.

## Mathematical Operations

```python
# r_1 + r_2
print("r_1:", r_1, "| shape:", r_1.shape)
print("r_2:", r_2, "| shape:", r_2.shape)
result = np.add(r_1, r_2)
print("r_1 + r_2:", result, "| shape:", result.shape)
```

```python
# r_1 + r_3
print("r_1:", r_1, "| shape:", r_1.shape)
print("r_3:", r_3, "| shape:", r_3.shape)
result = np.add(r_1, r_3)
print("r_1 + r_3:", result, "| shape:", result.shape)
```

```python
# r_2 + r_4
print("r_2:", r_2, "| shape:", r_2.shape)
print("r_4:", r_4, "| shape:", r_4.shape)
result = np.add(r_2, r_4)
print("r_2 + r_4:", result, "| shape:", result.shape)
```

```python
# What happens adding r_1, r_2, r_3 to r_4?
try:
    print(np.add(r_1, r_4))
except ValueError as e:
    print("r_1 + r_4 Error:", e)

try:
    print(np.add(r_2, r_4))
except ValueError as e:
    print("r_2 + r_4 Error:", e)

try:
    print(np.add(r_3, r_4))
except ValueError as e:
    print("r_3 + r_4 Error:", e)

# r_1 (shape 2,) cannot broadcast with r_4 (shape 4,5) — last dimensions don't match (2 vs 5)
# r_2 (shape 1,2) CAN broadcast with r_4 (shape 4,5) if dims align — but 2 != 5 so it errors
# r_3 (shape 2,1) CAN broadcast with r_4 (shape 4,5) — the 1 stretches across 5 columns, 2 rows stretch to 4
```

```python
# Add np.array([1]) to r_4 — broadcasts a single value to all elements
print(np.add(r_4, np.array([1])))
```

```python
# Add np.array([[1]]) to r_4 — same result, 2D scalar broadcasts to all
print(np.add(r_4, np.array([[1]])))
```

```python
# Add np.array([1,1,1,1,1]) to r_4 — broadcasts one row (5 elements) across all 4 rows
print(np.add(r_4, np.array([1, 1, 1, 1, 1])))
```

```python
# Add np.array([[1],[1]]) to r_4 — shape (2,1) can't broadcast to (4,5): 2 != 4
try:
    print(np.add(r_4, np.array([[1],[1]])))
except ValueError as e:
    print("Error:", e)
# Error: operands could not be broadcast together with shapes (4,5) and (2,1)
# The 2 rows in [[1],[1]] can't stretch to match 4 rows in r_4
```

### How does broadcasting work?

Broadcasting lets NumPy add arrays of different shapes by automatically "stretching"
the smaller array to match the larger one — but only when the dimensions are compatible.

The rules are:
- Dimensions are compared from right to left
- Two dimensions are compatible if they are **equal** or one of them is **1**
- A dimension of size 1 gets stretched (repeated) to match the other array's size

So `np.array([1])` added to a (4,5) array works because 1 broadcasts across everything.
`np.array([1,1,1,1,1])` works because shape (5,) matches the 5 columns.
The resulting shape is always the larger of the two arrays.

## Cumulative Operations

```python
# Print r_4 as starting values
print("r_4:")
print(r_4)
```

```python
# Sum of ALL values in r_4
print("Total sum:", np.sum(r_4))
```

```python
# Sum along axis=0 (sum each COLUMN across all rows)
print("Sum along axis=0 (column totals):", np.sum(r_4, axis=0))
```

```python
# Sum along axis=1 (sum each ROW across all columns)
print("Sum along axis=1 (row totals):", np.sum(r_4, axis=1))
```

```python
# Cumulative sum — rolling total of all values flattened
print("Cumulative sum (all values):")
print(np.cumsum(r_4))
```

```python
# Cumulative sum along axis=0 (rolling total down each column)
print("Cumulative sum along axis=0:")
print(np.cumsum(r_4, axis=0))
```

## Grocery Problem

```python
# bag1: [vegetables, fruits, snacks, beverages]
# bag1 = 2 vegetables, 3 fruits, 4 snacks, 0 beverages
# bag2 = 4 vegetables, 0 fruits, 1 snack, 2 beverages
bag1 = np.array([2, 3, 4, 0])
bag2 = np.array([4, 0, 1, 2])
print("bag1 [veg, fruit, snacks, bev]:", bag1)
print("bag2 [veg, fruit, snacks, bev]:", bag2)
```

```python
# Total items in each bag
bag1_total = np.sum(bag1)
bag2_total = np.sum(bag2)
print(f"The first bag contains {bag1_total} items, and the second bag contains {bag2_total} items.")
```

```python
# Total items per category across both bags
category_totals = np.add(bag1, bag2)
print(f"There are {category_totals[0]} vegetables, {category_totals[1]} fruits, "
      f"{category_totals[2]} boxes of snacks, and {category_totals[3]} beverages in the two grocery bags.")
```

## Teamwork

Scenario: Volunteer voter registration campaign. Three team members — Ali, Ben, and Carla —
knocked on doors Monday through Friday. Missing days are recorded as 0.

```python
# Ali: Mon=5, Tue=4, Wed=6, Thu=4, Fri=2
ali = np.array([5, 4, 6, 4, 2])
print("Ali:", ali)
```

```python
# Ben: Mon=0 (out of town), Tue=5, Wed=5, Thu=3, Fri=4
ben = np.array([0, 5, 5, 3, 4])
print("Ben:", ben)
```

```python
# Carla: Mon=3, Tue=6, Wed=0 (sick), Thu=3, Fri=4
carla = np.array([3, 6, 0, 3, 4])
print("Carla:", carla)
```

```python
# Compile all door-knocks into one 2D array (each row = one person)
all_knocks = np.array([ali, ben, carla])
print("all_knocks (rows = Ali, Ben, Carla | cols = Mon-Fri):")
print(all_knocks)
```

```python
# Total doors knocked per person (sum across axis=1 — each row)
person_totals = np.sum(all_knocks, axis=1)
print(f"Ali knocked on {person_totals[0]} doors total.")
print(f"Ben knocked on {person_totals[1]} doors total.")
print(f"Carla knocked on {person_totals[2]} doors total.")
print(f"Group total: {np.sum(all_knocks)} doors knocked.")
```

```python
# Rolling estimated votes turned out day by day
# Step 1: sum doors knocked per day (axis=0 = sum each column = daily totals)
daily_totals = np.sum(all_knocks, axis=0)
print("Daily door totals (Mon-Fri):", daily_totals)

# Step 2: cumulative rolling total by day
rolling_doors = np.cumsum(daily_totals)
print("Rolling total of doors knocked:", rolling_doors)

# Step 3: every 2 door-knocks = 1 estimated vote
estimated_votes = rolling_doors * 0.5
print("Rolling estimated votes turned out (Mon-Fri):", estimated_votes)
```
