```python
import numpy as np
import pandas as pd
```

## Basic Aggregations

```python
# Create df_sales
data = {
    'Product':    ['A', 'B', 'C', 'A', 'B', 'C', 'A', 'B', 'C'],
    'Region':     ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North'],
    'Units_Sold': [10, 15, 20, 5, 10, 15, 10, 5, 15],
    'Price':      [100, 200, 150, 100, 200, 150, 100, 200, 150]
}
df_sales = pd.DataFrame(data)
df_sales
```

```python
# Add Revenue column = Units_Sold * Price
df_sales['Revenue'] = df_sales['Units_Sold'] * df_sales['Price']
df_sales
```

```python
# Total Units_Sold and total Revenue
print("Total Units Sold:", df_sales['Units_Sold'].sum())
print("Total Revenue:   ", df_sales['Revenue'].sum())
```

```python
# Average Units_Sold and average Revenue
print("Average Units Sold:", df_sales['Units_Sold'].mean())
print("Average Revenue:   ", df_sales['Revenue'].mean())
```

## Grouping Data

```python
# Group by Product — total Units_Sold and Revenue per product
df_sales.groupby('Product')[['Units_Sold', 'Revenue']].sum()
```

```python
# Group by Region — average Units_Sold and Revenue per region
df_sales.groupby('Region')[['Units_Sold', 'Revenue']].mean()
```

```python
# Group by both Product AND Region — sum of Units_Sold and Revenue per combination
df_sales.groupby(['Product', 'Region'])[['Units_Sold', 'Revenue']].sum()
```

## Sorting and Ranking Data

```python
# Sort by Units_Sold ascending
df_sales.sort_values('Units_Sold', ascending=True)
```

```python
# Sort by Revenue descending
df_sales.sort_values('Revenue', ascending=False)
```

```python
# Add Revenue_Rank column — rank Revenue descending (rank 1 = highest revenue)
df_sales['Revenue_Rank'] = df_sales['Revenue'].rank(ascending=False)
df_sales
```

## Merging DataFrames

```python
# Create df_products
product_data = {
    'Product':  ['A', 'B', 'C'],
    'Category': ['Electronics', 'Furniture', 'Appliances']
}
df_products = pd.DataFrame(product_data)
df_products
```

```python
# Merge df_sales with df_products on Product column
df_merged = pd.merge(df_sales, df_products, on='Product')
df_merged
```

```python
# Total Revenue by Category
df_merged.groupby('Category')['Revenue'].sum()
```
