## Series

```python
# Create four plain Python objects
aa = [1, 2, 3, 4]                                              # list
ab = (5, 6, 7, 8)                                              # tuple
ac = {9, 10, 11, 12}                                           # set
ad = {'first': 13, 'second': 14, 'third': 15, 'fourth': 16}   # dictionary

print("aa (list):", aa)
print("ab (tuple):", ab)
print("ac (set):", ac)
print("ad (dict):", ad)
```

```python
# Create four pandas Series from the objects above
ser1 = pd.Series(aa)
ser2 = pd.Series(ab)
ser3 = pd.Series(ac)
ser4 = pd.Series(ad)

print("ser1 (from list):")
print(ser1)
print("\nser2 (from tuple):")
print(ser2)
print("\nser3 (from set):")
print(ser3)
print("\nser4 (from dict):")
print(ser4)
```

```python
# Try creating ser5 from literal values — this will raise a TypeError
try:
    ser5 = pd.Series(17, 18, 19, 20)
except TypeError as e:
    print("TypeError:", e)

# TypeError: __init__() takes from 1 to 2 positional arguments but 5 were given
# pd.Series() only accepts ONE positional argument (the data)
# To create a Series from literal values, wrap them in a list: pd.Series([17, 18, 19, 20])
```

## DataFrames from Series

```python
# Create df1 from ser1
df1 = pd.DataFrame(ser1)
df1
```

```python
# Add ser2 and ser3 as new columns using insert()
# insert(location, column_name, data)
df1.insert(1, 'ser2', ser2)
df1.insert(2, 'ser3', ser3)
df1
```

```python
# Add ser4 as a new column
df1['ser4'] = ser4
df1

# ser4 was created from a dictionary, so its INDEX is the dict keys ('first','second','third','fourth')
# df1's index is 0,1,2,3 (numeric) — these don't match ser4's string index
# Because the indices don't align, pandas can't match values and fills everything with NaN
```

```python
# Add a new row at index 4 using .loc
df1.loc[4] = [5, 9, 12, 0]
df1
```

### Info observations

From `df2.info()` we can determine:
- How many rows and columns the dataset has
- The data type of each column (int, float, object/string)
- How many non-null values each column has — any column with fewer non-null values than total rows has missing data
- How much memory the DataFrame is using

```python
# Column statistics for df2
df2.describe()
```

### Describe observations for sakila

**Useful insights:**
- `count` shows if any values are missing (if count < total rows)
- `mean`, `min`, `max` give a quick sense of the range of numeric columns
- `std` shows how spread out the values are

**Not so useful:**
- Stats on ID columns (like customer_id) are meaningless — the mean of an ID number tells us nothing
- `describe()` skips string/object columns entirely by default, so text data is excluded

```python
# Read northwind_example.csv into df3
df3 = pd.read_csv('northwind_example.csv')
df3
```

```python
# Column statistics for df3
df3.describe()
```

### Describe observations for northwind

**Useful insights:**
- Price and quantity columns give meaningful averages and ranges
- `min` and `max` help spot potential outliers or data entry errors

**How it differs from sakila:**
- Northwind is likely sales/order data so numeric columns like price and quantity are more meaningful to summarize
- Sakila is a movie rental database — many columns may be IDs or categorical text, making numeric stats less insightful
- The types of columns in each dataset determine how useful `describe()` is
