```python
import numpy as np
import matplotlib.pyplot as plt
```

## Creating Basic Plots

```python
# Data points
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]
```

```python
# Basic line graph
plt.plot(x, y)
plt.show()
```

```python
# Line graph with circle markers
plt.plot(x, y, marker='o')
plt.show()
```

```python
# Add title and axis labels
plt.plot(x, y, marker='o')
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

## Styling Plots

```python
# Dashed red line
plt.plot(x, y, linestyle='--', color='red')
plt.title('Dashed Red Line')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

```python
# Square markers, size 10, green marker color
plt.plot(x, y, linestyle='--', color='red', marker='s', markersize=10, markerfacecolor='green')
plt.title('Square Green Markers')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

```python
# Experiment 1: triangle marker ('^'), solid blue line, purple markers
plt.plot(x, y, linestyle='-', color='blue', marker='^', markersize=12, markerfacecolor='purple')
plt.title('Triangle Purple Markers')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

```python
# Experiment 2: diamond marker ('D'), dotted orange line, cyan markers
plt.plot(x, y, linestyle=':', color='orange', marker='D', markersize=10, markerfacecolor='cyan')
plt.title('Diamond Cyan Markers')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```
