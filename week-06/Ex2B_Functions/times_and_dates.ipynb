# Exercise 2.B Using library functions
## Lab 2. Working with datetime functions

```python
# Import the datetime module
import datetime
```

# Today

```python
# Define today as the current date and time
today = datetime.datetime.now()
```

```python
# Format and print today's date
# %A = full day name, %B = full month name, %d = day, %Y = 4-digit year
print(today.strftime("Today is %A, %B %d, %Y"))
```

# My birthday

```python
# Define a birthday using datetime.date(year, month, day)
birthday = datetime.date(1998, 7, 15)
```

```python
# Format and print the birthday as MM/DD/YYYY
# %m = zero-padded month, %d = zero-padded day, %Y = 4-digit year
print(birthday.strftime("My birthday is %m/%d/%Y"))
```

# Next 90 Days

```python
# Calculate the date 90 days from today using timedelta
ninety_d = today + datetime.timedelta(days=90)
```

```python
# Format and print the date 90 days from now
print(ninety_d.strftime("90 days from today is %B %d, %Y"))
```

# Time for Dinner

```python
# Define a dinner time using datetime.time(hour, minute)
# Using 7:30 PM (19:30 in 24-hour format)
dinner_time = datetime.time(19, 30)
```

```python
# Format and print the dinner time in 12-hour format
# %I = 12-hour clock, %M = minutes, %p = AM/PM
print(dinner_time.strftime("Let's meet for dinner at %I:%M %p"))
```
