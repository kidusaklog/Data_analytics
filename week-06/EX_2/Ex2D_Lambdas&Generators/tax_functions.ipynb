## Lab 3. Payroll Tax Functions

This notebook demonstrates how to use custom functions with conditional logic to calculate payroll tax withholdings.

Three functions are defined:
- `get_soc_sec_tax()` — calculates Social Security tax at 6.2%
- `get_medicare_tax()` — calculates Medicare tax at 1.45%
- `get_federal_tax()` — calculates federal withholding based on gross pay and a withholding code (0–4+)

```python
# Social Security tax: 6.2% of gross pay
def get_soc_sec_tax(gross_pay):
    rate = 0.062
    return gross_pay * rate
```

```python
# Medicare tax: 1.45% of gross pay
def get_medicare_tax(gross_pay):
    rate = 0.0145
    return gross_pay * rate
```

```python
# Federal withholding tax based on gross pay and withholding code
# Withholding code rates:
#   0  -> 23%
#   1  -> 21%
#   2  -> 19.5%
#   3  -> 18.5%
#   4+ -> 18%, minus 0.5% for each withholding OVER 4
#         e.g. code 6: 0.18 - (2 x 0.005) = 0.17 (17%)
def get_federal_tax(gross_pay, withholding_code):
    if withholding_code == 0:
        rate = 0.23
    elif withholding_code == 1:
        rate = 0.21
    elif withholding_code == 2:
        rate = 0.195
    elif withholding_code == 3:
        rate = 0.185
    elif withholding_code == 4:
        rate = 0.18
    else:
        # For codes above 4, subtract 0.5% for each step over 4
        extra = withholding_code - 4
        rate = 0.18 - (extra * 0.005)
    return gross_pay * rate
```

```python
# Person 1: gross pay $750, withholding code 0
gross1 = 750
code1  = 0

ss1      = get_soc_sec_tax(gross1)
med1     = get_medicare_tax(gross1)
federal1 = get_federal_tax(gross1, code1)

print(f"Person 1 — Gross Pay: ${gross1:.2f} | Withholding Code: {code1}")
print(f"  Social Security Tax : ${ss1:.2f}")
print(f"  Medicare Tax        : ${med1:.2f}")
print(f"  Federal Tax         : ${federal1:.2f}")
print(f"  Total Withheld      : ${ss1 + med1 + federal1:.2f}")
```
