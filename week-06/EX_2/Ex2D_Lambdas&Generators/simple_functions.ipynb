# Exercise 2.C Creating and calling functions
## Lab 1. Working with defining custom functions
In this lab, we practice creating simple functions with parameters and return values, then calling them with arguments to produce output.

```python
# Function 1: favorite_things
# Parameters: name, movie, music
# Returns 3 lines listing name, favorite movie, and favorite musician/band
def favorite_things(name, movie, music):
    return (f"Name: {name}\n"
            f"Favorite Movie: {movie}\n"
            f"Favorite Musician/Band: {music}")

print(favorite_things('Alex Johnson', 'The Matrix', 'Kendrick Lamar'))
```

```python
# Function 2: why_im_here
# Parameter: purpose
# Returns the purpose sentence describing why the person joined Year Up United
def why_im_here(purpose):
    return purpose

print(why_im_here('I joined the Year Up United Data Analyst program to build technical skills '
                  'that will help me launch a career in the tech industry.'))
```

```python
# Function 3: favorite_place
# Parameters: place, why
# Returns a sentence describing a favorite place and why it is a favorite
def favorite_place(place, why):
    return f"One of my favorite places is {place} because {why}."
```

```python
# Test favorite_place with first example
print(favorite_place('Chicago', 'the food scene and architecture are incredible'))
```

```python
# Test favorite_place with a different place and reason
print(favorite_place('the beach', 'it helps me relax and clear my mind'))
```
