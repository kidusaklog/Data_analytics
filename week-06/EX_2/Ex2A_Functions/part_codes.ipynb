# Exercise 2.A Using built-in functions
## Lab 2. Parsing strings – part codes

```python
# get_supplier_code: returns all characters before the colon (:)
def get_supplier_code(part_code):
    # Find the position of the colon, then slice everything before it
    colon_index = part_code.index(':')
    return part_code[:colon_index]
```

```python
# get_product_num: returns all characters between the colon (:) and the dash (-)
def get_product_num(part_code):
    # Find the positions of both the colon and dash, then slice between them
    colon_index = part_code.index(':')
    dash_index = part_code.index('-')
    return part_code[colon_index+1:dash_index]
```

```python
# get_size: returns all characters after the dash (-)
def get_size(part_code):
    # Find the position of the dash, then slice everything after it
    dash_index = part_code.index('-')
    return part_code[dash_index+1:]
```

```python
# Variables for the three example part codes
code1 = 'ACME:123-L'
code2 = 'DI:12345-M'
code3 = 'ACE:1-12'
```

```python
# Test all three functions with code1: ACME:123-L
supplier = get_supplier_code(code1)
product  = get_product_num(code1)
size     = get_size(code1)
print(f"For {code1}, the supplier is {supplier}. Part number {product} is available in size {size}.")
```

```python
# Test all three functions with code2: DI:12345-M
supplier = get_supplier_code(code2)
product  = get_product_num(code2)
size     = get_size(code2)
print(f"For {code2}, the supplier is {supplier}. Part number {product} is available in size {size}.")
```

```python
# Test all three functions with code3: ACE:1-12
supplier = get_supplier_code(code3)
product  = get_product_num(code3)
size     = get_size(code3)
print(f"For {code3}, the supplier is {supplier}. Part number {product} is available in size {size}.")
```

```python
# BONUS: Store all part codes in a list and loop through them
part_codes_list = ['ACME:123-L', 'DI:12345-M', 'ACE:1-12']

for code in part_codes_list:
    supplier = get_supplier_code(code)
    product  = get_product_num(code)
    size     = get_size(code)
    print(f"For {code}, the supplier is {supplier}. Part number {product} is available in size {size}.")
```
