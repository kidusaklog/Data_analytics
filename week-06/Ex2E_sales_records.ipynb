This notebook analyzes a small sales dataset using Python functions.
Each function takes the full dataset as its first argument.

```python
# Full sales dataset: each tuple contains (product, region, sale_amount)
sales_data = [
    ('Laptop',   'East',  1200.00),
    ('Phone',    'West',   850.00),
    ('Tablet',   'East',   400.00),
    ('Monitor',  'South',  650.00),
    ('Keyboard', 'West',    75.00),
    ('Laptop',   'South', 1150.00),
    ('Phone',    'East',   900.00),
    ('Tablet',   'West',   425.00),
    ('Monitor',  'East',   680.00),
    ('Keyboard', 'South',   80.00),
    ('Laptop',   'West',  1300.00),
    ('Phone',    'West',   820.00),
]
```

## Lab 1. Total sales by region

```python
# region_total: returns the sum of all sale amounts for a given region
# Uses a generator expression to filter by region and sum the amounts
# The underscore _ is a Python convention for an unpacked value we don't need
def region_total(data, region):
    return sum(amt for _, r, amt in data if r == region)
```

```python
# Test region_total for all three regions
print(region_total(sales_data, 'East'))   # expected: 3180.00
print(region_total(sales_data, 'West'))   # expected: 3450.00
print(region_total(sales_data, 'South'))  # expected: 1880.00
```

## Lab 2. Count of sales by region

```python
# region_count: returns the number of records for a given region
# Same generator pattern as region_total, but sums 1 per matching record
def region_count(data, region):
    return sum(1 for _, r, _ in data if r == region)
```

```python
# Test region_count for all three regions
print(region_count(sales_data, 'East'))   # expected: 4
print(region_count(sales_data, 'West'))   # expected: 5
print(region_count(sales_data, 'South'))  # expected: 3
```

## Lab 3. Average sale amount by region

```python
# region_average: returns the average sale amount for a given region
# Calls region_total() and region_count() rather than re-writing the logic
# Returns 0 if there are no records for that region (avoids division by zero)
def region_average(data, region):
    count = region_count(data, region)
    if count == 0:
        return 0
    return region_total(data, region) / count
```

```python
# Test region_average for all three regions, plus 'North' which has no records
print(region_average(sales_data, 'East'))   # expected: 795.00
print(region_average(sales_data, 'West'))   # expected: 690.00
print(region_average(sales_data, 'South'))  # expected: ~626.67
print(region_average(sales_data, 'North'))  # expected: 0 (no records)
```

## Lab 4. Filter by sale amount

```python
# sales_above: returns a new list of records where the sale amount > threshold
def sales_above(data, threshold):
    return [record for record in data if record[2] > threshold]
```

```python
# Test sales_above with a threshold of 500 - print each matching record
above_500 = sales_above(sales_data, 500)
print("Sales above $500:")
for record in above_500:
    print(record)
```

```python
# BONUS: sales_below - returns records where the sale amount is less than the threshold
def sales_below(data, threshold):
    return [record for record in data if record[2] < threshold]

below_500 = sales_below(sales_data, 500)
print("Sales below $500:")
for record in below_500:
    print(record)
```

## Lab 5. Top and bottom sale

```python
# top_sale: returns the record (tuple) with the highest sale amount
# bottom_sale: returns the record (tuple) with the lowest sale amount
# Both use the key= argument with a lambda to compare by sale amount (index 2)
def top_sale(data):
    return max(data, key=lambda s: s[2])

def bottom_sale(data):
    return min(data, key=lambda s: s[2])

print("Top sale:   ", top_sale(sales_data))
print("Bottom sale:", bottom_sale(sales_data))
```

The top sale was a **Laptop** in the **West** region for **$1,300.00**.

The bottom sale was a **Keyboard** in the **West** region for **$75.00**.

## Lab 6. Apply a discount

```python
# apply_discount: returns a NEW list with each sale amount reduced by discount_pct
# Product name and region remain unchanged; only the amount is updated
# Uses map() with a lambda to transform every record, wrapped in list()
def apply_discount(data, discount_pct):
    return list(map(lambda s: (s[0], s[1], s[2] * (1 - discount_pct)), data))
```

```python
# Test apply_discount at 15% - compare the first 3 records before and after
discounted = apply_discount(sales_data, 0.15)

print("First 3 records BEFORE 15% discount:")
for record in sales_data[:3]:
    print(record)

print("\nFirst 3 records AFTER 15% discount:")
for record in discounted[:3]:
    print(record)
```

## Lab 7. Unique regions

```python
# unique_regions: returns a sorted list of unique region names in the dataset
# set() automatically removes duplicates; sorted() puts them in alphabetical order
def unique_regions(data):
    return sorted(set(r for _, r, _ in data))

print(unique_regions(sales_data))
```

**What is `unique_regions()` useful for?**

This pattern is useful any time you need to discover the distinct categories in a dataset before doing analysis — for example, getting a list of all departments in an HR file, all product categories in an inventory, or all states in a customer list. Instead of hardcoding those values, you can call `unique_regions()` (or a similar function) to build them dynamically. This makes your code flexible and reusable: if a new region is added to the data, your analysis will automatically include it without any code changes.

## Lab 8. Region summary

```python
# region_summary: prints a formatted report for one region
# Calls region_count(), region_total(), region_average(), and top_sale()
# Filters the dataset down to the target region before calling top_sale()
def region_summary(data, region):
    # Narrow dataset to just this region so top_sale() compares within region only
    region_data = [record for record in data if record[1] == region]

    count   = region_count(data, region)
    total   = region_total(data, region)
    average = region_average(data, region)
    top     = top_sale(region_data)  # top sale within this region

    print(f"=== {region} Region ===")
    print(f"Records:      {count}")
    print(f"Total Sales:  ${total:,.2f}")
    print(f"Average Sale: ${average:,.2f}")
    print(f"Top Sale:     {top[0]} — ${top[2]:,.2f}")
    print()
```

```python
# Test region_summary with the East region
region_summary(sales_data, 'East')
```

## Lab 9. Full summary

```python
# Loop through every unique region and print a summary for each
# unique_regions() provides the list dynamically - no hardcoded region names needed
for region in unique_regions(sales_data):
    region_summary(sales_data, region)
```
