```python
# Import NumPy
import numpy as np
```

## Array Basics

```python
# 5a - Create a standard Python list named fruits
fruits = ["apple", "banana", "cherry", "date"]
print(fruits)
```

```python
# 5b - Create arr_1 as a 1D NumPy array from fruits
arr_1 = np.array(fruits)
print(arr_1)
```

```python
# 5c - Check the data type of values in arr_1
print(arr_1.dtype)

# The dtype returned is '<U6'
# The letter 'U' stands for Unicode string (text data)
# The number '6' is the max number of characters in the longest string in the array
# ('banana' and 'cherry' both have 6 characters, so NumPy allocates 6 character slots)
```

```python
# 5d - Create arr_2 as a 2D array containing (10,11,12) and (13,14,15)
arr_2 = np.array([(10, 11, 12), (13, 14, 15)])
print(arr_2)
```

```python
# 5e - Print arr_2 with dimensions transposed using .T
print(arr_2.T)
```

```python
# Does .T actually change the value stored in arr_2?
print(arr_2)

# No! arr_2.T does NOT alter the original array.
# .T returns a VIEW of the transposed array without modifying arr_2 itself.
# To save the transposed version you would need: arr_2_transposed = arr_2.T
```

```python
# 5f - Create arr_3 as a 2D array containing (1,3,5) and (2,4,6)
arr_3 = np.array([(1, 3, 5), (2, 4, 6)])
print(arr_3)
```

```python
# 5g - Create arr_4 using slicing from arr_3
# Element at index [0, 2] = row 0, column 2 = 5
# Element at index [1, 1] = row 1, column 1 = 4
arr_4 = np.array([arr_3[0, 2], arr_3[1, 1]])
print(arr_4)
```

```python
# 5h - Create arr_5 as a 3D array of all 1's with 2 values in each dimension
# Shape = (2, 2, 2) means 2 blocks, 2 rows, 2 columns
arr_5 = np.ones((2, 2, 2))
print(arr_5)
```
