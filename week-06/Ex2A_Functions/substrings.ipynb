## Lab 1. Working with names – parsing strings

```python
# Define a test name variable
test_name = 'John Doe'
```

```python
# Function to parse a full name into first and last, then display results
def parse_and_display(name):
    for i in range(len(name)):
        if name[i] == ' ':
            first = name[:i]
            last = name[i+1:]
    print(f"Full name: {name}")
    print(f"First name: {first}")
    print(f"Last name: {last}")
```

```python
# Call the function with the test_name variable
parse_and_display(test_name)
```

```python
# List of names to test
test_list = ['Grace Flores', 'JB Oinonen', 'Ken Penn']
```

```python
# Test with Grace Flores
parse_and_display(test_list[0])
```

```python
# Test with JB Oinonen
parse_and_display(test_list[1])
```

```python
# Test with Ken Penn
parse_and_display(test_list[2])
```

```python
# Test with your own name entered directly as a string
parse_and_display('Alex Johnson')
```

```python
# CHALLENGE: parse_and_display2 handles one, two, or three word names
def parse_and_display2(name):
    # Split the name into parts using the built-in split() method
    parts = name.split(' ')
    
    if len(parts) == 1:
        # Only one word - treat it as a single/mononym
        print(f"Full name: {name}")
        print(f"Name: {parts[0]}")
        print(f"(No first/last name separation - single name)")
    
    elif len(parts) == 2:
        # Two words - first and last name
        first = parts[0]
        last = parts[1]
        print(f"Full name: {name}")
        print(f"First name: {first}")
        print(f"Last name: {last}")
    
    elif len(parts) == 3:
        # Three words - first, middle, and last name
        first = parts[0]
        middle = parts[1]
        last = parts[2]
        print(f"Full name: {name}")
        print(f"First name: {first}")
        print(f"Middle name: {middle}")
        print(f"Last name: {last}")
    
    else:
        print(f"Name '{name}' has more than 3 parts - unable to parse.")
    
    print()  # blank line for readability between results
```

```python
# Test parse_and_display2 with one, two, and three word names
parse_and_display2('Lorde')
parse_and_display2('Billie Eilish')
parse_and_display2('Megan Thee Stallion')
```
