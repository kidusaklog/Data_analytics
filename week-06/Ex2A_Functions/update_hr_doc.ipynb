## Lab 3. Employee data
This lab exercise makes updates to information extracted from an employee database.

```python
# Define the HR list of employee tuples: (ID, Department, Name, Salary)
hr_list = [
    ('0123', 'HR',  'Rebecca Yang',  '69000'),
    ('0121', 'IT',  'Mark Blick',    '67000'),
    ('0068', 'IT',  'Kahna Larsen',  '112000'),
    ('0234', 'OPS', 'Jim Smith',     '54000')
]
```

```python
# UPDATE 1: Mark reported a name change - update last name to Blick-Hawley
# Tuples are immutable, so we replace the entire tuple at the correct index
for i in range(len(hr_list)):
    if hr_list[i][2] == 'Mark Blick':
        # Rebuild the tuple with the updated name
        hr_list[i] = (hr_list[i][0], hr_list[i][1], 'Mark Blick-Hawley', hr_list[i][3])
```

```python
# Print the list after Update 1
print("After Update 1 (Mark's name change):")
for employee in hr_list:
    print(employee)
```

```python
# UPDATE 2: Jim accepted a new role - change department from OPS to CS, salary to 60000
for i in range(len(hr_list)):
    if hr_list[i][2] == 'Jim Smith':
        # Rebuild the tuple with the updated department and salary
        hr_list[i] = (hr_list[i][0], 'CS', hr_list[i][2], '60000')
```

```python
# Print the list after Update 2 (both changes should persist)
print("After Update 2 (Jim's department and salary change):")
for employee in hr_list:
    print(employee)
```

```python
# Display the final list in formatted output:
# Employee# | DeptCode | Name | NN,NNN
print("Final Employee List:")
print("-" * 50)
for emp in hr_list:
    emp_id   = emp[0]
    dept     = emp[1]
    name     = emp[2]
    # Cast salary to int first so format() can apply comma formatting
    salary   = int(emp[3])
    print(f"{emp_id} | {dept} | {name} | {salary:,}")
```
